```python
students = {}

def add():
    n = int(input("Enter how many students want to add: "))

    for i in range(n):
        name = input("Enter a name: ")

        try:
            marks = int(input("Enter marks: "))
            students[name] = marks

        except ValueError:
            print("Please enter marks as a number!")


def delete():
    found = False
    delete_name = input("Enter the name you want to delete: ")

    for name in list(students.keys()):
        if name.lower() == delete_name.lower():
            del students[name]
            print(f"{delete_name} is deleted successfully")
            found = True
            break

    if not found:
        print("The name is not in dictionary")


def update():
    choice = input("What do you want to update (name/marks)?: ")
    search_name = input("Enter the student name: ")
    found = False

    for student, marks in students.items():

        if student.lower() == search_name.lower():

            if choice.lower() == "marks":
                new_marks = int(input("Enter new marks: "))
                students[student] = new_marks
                print(f"{student}'s marks updated to {new_marks}")

            elif choice.lower() == "name":
                new_name = input("Enter new name: ")
                students[new_name] = students.pop(student)
                print(f"Name updated from {student} to {new_name}")

            else:
                print("Invalid update choice!")

            found = True
            break

    if not found:
        print("Student not found")


def view():
    print("------ Students ------")
    for name, marks in students.items():
        print(f"{name}: {marks}")


def average():

    if students:
        total = 0

        for marks in students.values():
            total += marks
        avg = total / len(students)
        print("Total marks:", total)
        print("Average marks:", avg)

    else:
        print("No students available")


def topper():

    if students:
        topper_name = max(students, key=students.get)

        print(
            f"Topper is {topper_name} "
            f"with {students[topper_name]} marks"
        )

    else:
        print("No students available")


while True:

    print("\n------ Student Management System ------")
    print("1. Add student")
    print("2. Update student")
    print("3. Delete")
    print("4. View")
    print("5. Average")
    print("6. Topper")
    print("7. Exit")

    user_input = int(input("Enter the choice: "))

    if user_input == 1:
        add()

    elif user_input == 2:
        update()

    elif user_input == 3:
        delete()

    elif user_input == 4:
        view()

    elif user_input == 5:
        average()

    elif user_input == 6:
        topper()

    elif user_input == 7:
        print("------- Exiting --------")
        break

    else:
        print("Invalid choice!")
```
